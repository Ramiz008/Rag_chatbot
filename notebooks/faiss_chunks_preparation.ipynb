```python
from sentence_transformers import SentenceTransformer
import json
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
with open("chunks/chunks.json", "r") as f:
    chunks = json.load(f)

embeddings =model.encode(chunks)
```

```python
import faiss
```

```python
# Create FAISS index
Findex = faiss.IndexFlatL2(embeddings.shape[1])
Findex.add(embeddings)
```

```python
faiss.write_index(Findex,"vectordb/faiss.index")
```
